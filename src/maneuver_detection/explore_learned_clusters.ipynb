## ce notebook a pour but d'explorer les params orbitaux des clusters appris par apprentissage non supervisé sur les données spacetrack

```python
import os
import pandas as pd
from pathlib import Path
import csv

from viz.time_series import plot_cluster, plot_time_series, plot_compare

base = Path.cwd()
```

```python
spacetrack_data_dir = Path(os.path.join(base , '..', '..', 'data', 'raw', 'spacetrack', 'max_objects_all_regimes'))


dataframes = sorted(spacetrack_data_dir.glob("*.parquet"))

df_TLE = pd.concat([pd.read_parquet(p) for p in dataframes], ignore_index = True)
```

```python

import matplotlib.pyplot as plt
import re 
import gc 

CLUSTER_DIR = os.path.join(base, '..', '..', 'outputs', 'clustering' , 'cluster_leo_15000_128D_MCS15_MS15_OPTIM_named.csv')
figures_dir = Path(os.path.join(base, '..', '..', 'outputs', 'clustering'))

def sanitize_filename(filename):
    """Remplace les caractères interdits dans un nom de fichier."""
    # Garde seulement alphanumérique, espace, underscore, tiret
    return re.sub(r'[^\w\-_ ]', '_', filename)


def learned_clusters_params(cluster_dir, df_tle) : 

    df_cluster = pd.read_csv(cluster_dir)
    valid = df_cluster[df_cluster['cluster'] != -1]
    norads = valid['norad']

    df_tle_valid= df_tle[df_tle['norad'].isin(norads)]
    
    for cluster_id, sub in valid.groupby('cluster') : 
        norads = sub['norad']
        cluster_name = sub['cluster_name'].iloc[0] 
        cluster_label = f'Cluster {cluster_id} : {len(norads)} objets, {cluster_name}'
        fig = plot_cluster(
            df_tle_valid, 
            norads, 
            ['sma' , 'inclination' , 'eccentricity', 'raan', 'arg_perigee', 'mean_anomaly'], 
            cluster_label=cluster_label
            )
        label = sanitize_filename(cluster_label)
        fig_name = figures_dir / f"{label}.png"
        fig.savefig(fig_name)
        print(f"fichier {fig_name}")
        plt.close(fig)
        del fig, norads, sub, cluster_name, cluster_label, label
        gc.collect()  # Force le garbage collector

learned_clusters_params(CLUSTER_DIR, df_TLE)
```
